```python
# [CELL 16] ADDITIONAL EXERCISE 1: EMERGENCY DISPATCH RECORD SEARCH

# Student identity setup
LAST_NAME = "Dela Cruz"
STUDENT_ID = "TUPM-26-1251"
FAVORITE_ARTIST = "ILLIT"

# Calculate deterministic generation variables
SEED_DIGIT = int(STUDENT_ID[-1])                      # 1
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit()) # 17
NAME_LENGTH = len(LAST_NAME)                          # 9
ARTIST_LENGTH = len(FAVORITE_ARTIST)                  # 5
ARTIST_SUM = sum(ord(c) - ord('A') + 1 for c in FAVORITE_ARTIST.upper()) # 55

# 1. Dataset Generation Rules
# Generate an ordered sequence of unique incident IDs
base_dataset = [100 + i * 7 + (ID_SUM % 5) for i in range(15)]

# Rotate the dataset at a pivot point determined by SEED_DIGIT
rotation_pivot = (SEED_DIGIT % len(base_dataset)) if SEED_DIGIT > 0 else 4
rotated_dataset = base_dataset[rotation_pivot:] + base_dataset[:rotation_pivot]

# Generate a deterministic target search parameter
target_incident = 100 + ((ID_SUM + ARTIST_SUM) % len(base_dataset)) * 7 + (ID_SUM % 5)

print("Original Dataset :", base_dataset)
print("Rotated Dataset  :", rotated_dataset)
print("Target Incident  :", target_incident)


# 2. Rotated Binary Search Algorithm
def rotated_binary_search(data, target):
    low, high, comparisons = 0, len(data) - 1, 0
    while low <= high:
        mid = (low + high) // 2
        comparisons += 1
        
        if data[mid] == target:
            return mid, comparisons
            
        # Left half is properly sorted
        if data[low] <= data[mid]:
            if data[low] <= target < data[mid]:
                high = mid - 1
            else:
                low = mid + 1
        # Right half is properly sorted
        else:
            if data[mid] < target <= data[high]:
                low = mid + 1
            else:
                high = mid - 1
                
    return -1, comparisons


# 3. Baseline Linear Search Algorithm
def linear_search_comparison(data, target):
    comparisons = 0
    for i, value in enumerate(data):
        comparisons += 1
        if value == target:
            return i, comparisons
    return -1, comparisons


# 4. Processing & Execution Report
b_index, b_comps = rotated_binary_search(rotated_dataset, target_incident)
l_index, l_comps = linear_search_comparison(rotated_dataset, target_incident)

print("\nEXECUTION PERFORMANCE REPORT")
print(f"Target Incident ID      : {target_incident}")
print(f"Rotated Index Location  : {b_index}")
print(f"Linear Search Comps     : {l_comps}")
print(f"Rotated Binary Comps    : {b_comps}")
print(f"Complexity Profile      : Time: O(log n) | Space: O(1)")
```

```python
# [CELL 17] ADDITIONAL EXERCISE 2: UNIVERSITY ENROLLMENT RECORDS

# Student identity setup
LAST_NAME = "Dela Cruz"
STUDENT_ID = "TUPM-26-1251"
FAVORITE_ARTIST = "ILLIT"

# Calculate deterministic generation variables
SEED_DIGIT = int(STUDENT_ID[-1])                      # 1
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit()) # 17
NAME_LENGTH = len(LAST_NAME)                          # 9
ARTIST_LENGTH = len(FAVORITE_ARTIST)                  # 5

# 1. Dataset Generation Rules (Sorted Department Arrays)
dept_A = [2000 + (SEED_DIGIT * 3) + i * (ID_SUM % 5 + 2) for i in range(8)]
dept_B = [2000 + (ARTIST_LENGTH * 2) + i * (NAME_LENGTH % 4 + 3) for i in range(7)]

print("Department A dataset:", dept_A)
print("Department B dataset:", dept_B)


# 2. Optimized Two-Pointer Consolidation (Merge) Algorithm
def consolidate_records(list_a, list_b):
    i, j = 0, 0
    consolidated = []
    comparisons = 0
    
    while i < len(list_a) and j < len(list_b):
        comparisons += 1
        if list_a[i] <= list_b[j]:
            consolidated.append(list_a[i])
            i += 1
        else:
            consolidated.append(list_b[j])
            j += 1
            
    # Append any remaining sorted elements directly (no comparisons needed)
    while i < len(list_a):
        consolidated.append(list_a[i])
        i += 1
        
    while j < len(list_b):
        consolidated.append(list_b[j])
        j += 1
        
    return consolidated, comparisons


# 3. Processing and Execution
consolidated_list, merge_comparisons = consolidate_records(dept_A, dept_B)

# Verify structural integrity and ordering
records_before = len(dept_A) + len(dept_B)
records_after = len(consolidated_list)
is_sorted = consolidated_list == sorted(consolidated_list)

print("\nCONSOLIDATION PERFORMANCE REPORT")
print("Consolidated dataset :", consolidated_list)
print(f"Records Before       : {records_before}")
print(f"Records After        : {records_after}")
print(f"Comparison Count     : {merge_comparisons}")
print(f"Ordering Verification: {'PASSED' if is_sorted else 'FAILED'}")
print(f"Complexity Profile   : Time: O(n + m) | Space: O(n + m)")
```

```python
# [CELL 18] ADDITIONAL EXERCISE 3: E-COMMERCE ORDER PROCESSING
import time

# Student identity setup
LAST_NAME = "Dela Cruz"
STUDENT_ID = "TUPM-26-1251"

# Calculate deterministic generation variables
SEED_DIGIT = int(STUDENT_ID[-1])                      # 1
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit()) # 17
NAME_LENGTH = len(LAST_NAME)                          # 9

# 1. Base Core Sorting Functions with Trackers
def tracked_bubble_sort(data_input):
    a, comparisons, movements = data_input.copy(), 0, 0
    n = len(a)
    for i in range(n):
        swapped = False
        for j in range(n - i - 1):
            comparisons += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                movements += 3 # A standard swap requires 3 data value assignments
                swapped = True
        if not swapped:
            break
    return a, comparisons, movements

def tracked_selection_sort(data_input):
    a, comparisons, movements = data_input.copy(), 0, 0
    n = len(a)
    for i in range(n - 1):
        minimum = i
        for j in range(i + 1, n):
            comparisons += 1
            if a[j] < a[minimum]:
                minimum = j
        if minimum != i:
            a[i], a[minimum] = a[minimum], a[i]
            movements += 3 # A standard swap requires 3 data value assignments
    return a, comparisons, movements

def tracked_insertion_sort(data_input):
    a, comparisons, movements = data_input.copy(), 0, 0
    for i in range(1, len(a)):
        key = a[i]
        movements += 1
        j = i - 1
        while j >= 0:
            comparisons += 1
            if a[j] <= key:
                break
            a[j + 1] = a[j]
            movements += 1
            j -= 1
        a[j + 1] = key
        movements += 1
    return a, comparisons, movements

# 2. Benchmarking Controller Function
def profile_sorting_algorithm(algorithm_func, dataset):
    start_time = time.perf_counter()
    sorted_res, comps, moves = algorithm_func(dataset)
    elapsed_time = time.perf_counter() - start_time
    verification = "PASSED" if sorted_res == sorted(dataset) else "FAILED"
    return comps, moves, elapsed_time, verification

# 3. Data Generation (Orders Dataset Creation)
# Generates 20 deterministic numbers using student inputs
base_orders = [(100 + (i * ID_SUM + SEED_DIGIT * NAME_LENGTH) % 250) for i in range(20)]

# Establish the three explicit input states
input_conditions = {
    "Sorted": sorted(base_orders),
    "Reverse-Ordered": sorted(base_orders, reverse=True),
    "Random": base_orders.copy()
}

print("Generated Order Dataset (Random Baseline):", base_orders)

# 4. Run Benchmarking Matrix & Present Report Tables
sorting_algorithms = {
    "Bubble Sort": tracked_bubble_sort,
    "Selection Sort": tracked_selection_sort,
    "Insertion Sort": tracked_insertion_sort
}

print("\nE-COMMERCE BENCHMARKING REPORT")
for condition_name, text_dataset in input_conditions.items():
    print(f"\nCondition State: {condition_name}")
    print(f"{'Algorithm':<16}{'Comparisons':>12}{'Movements':>12}{'Time (s)':>15}{'Verification':>14}")
    
    for algo_name, algo_func in sorting_algorithms.items():
        comps, moves, duration, verify = profile_sorting_algorithm(algo_func, text_dataset)
        print(f"{algo_name:<16}{comps:>12}{moves:>12}{duration:>15.8f}s{verify:>14}")
```
